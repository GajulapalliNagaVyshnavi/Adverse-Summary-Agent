```python

client = OpenAI(
	api_key=os.getenv("OPENAI_API_KEY"),
)
```

```python
def load_data(file_path):
    return pd.read_csv(file_path)
```

```python
def preprocess_data(df):
    # Remove any empty rows or columns
    df = df.dropna(how='all').dropna(axis=1, how='all')
    return df
```

```python
def generate_summary(data):
    prompt = f"""
    
You will be given data tables that summarize treatment-emergent adverse events (AEs) reported in a clinical trial comparing varenicline to placebo. Your task is to summarize the key information regarding adverse events and relate it to the study's objectives.

Context: The study aimed to compare the efficacy of varenicline 1 mg BID to placebo for smoking cessation in subjects with cardiovascular disease over a 12-week treatment period, followed by a 40-week follow-up. Safety data were gathered during this period.

Instructions:

Summarize the total number of subjects and the overall findings related to treatment-emergent adverse events (TEAEs) for both the varenicline and placebo groups.
Highlight any adverse events that occurred in greater than 2% of subjects in the varenicline group and were more frequent than in the placebo group.
Discuss the types of adverse events reported by system organ class, focusing on the most common ones.
Provide the total number and percentage of subjects who experienced any adverse events in both treatment arms.
Identify any serious adverse events (SAEs) reported and their relevance to the treatment.
Analyze any trends or patterns in the data, such as whether specific AEs are associated more frequently with varenicline than with placebo.
Relate these findings back to the study's safety objectives, specifically regarding subjects with cardiovascular disease.
Conclude by assessing the overall safety profile of varenicline in comparison to placebo, discussing the implications for its use in smoking cessation therapy in this patient population.

    Input tables:

    {data.to_string()}

    Please provide a summary based on the above instructions and data.
    """

    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": "You are a clinical trial data analyst specializing in summarizing adverse event data."},
            {"role": "user", "content": prompt}
        ],
        max_tokens=1000
    )

  #  return response.choices[0].message.content
    summary = response.choices[0].message.content
    
    # Clean up the summary to remove hashtags and stars
    clean_summary = summary.replace('*', '').replace('#', '')
    
    return clean_summary
```

```python

file_path = 'adverse data.csv'
df = load_data(file_path)
processed_df = preprocess_data(df)
```

```python
print("Processed Data:")
display(processed_df)
```

```python
generated_summary = generate_summary(processed_df)
print("\nGenerated Summary:")
print(generated_summary)
```
