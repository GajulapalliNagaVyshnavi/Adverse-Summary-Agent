```python

import pandas as pd
import os
from openai import OpenAI
import plotly.express as px
from IPython.display import display, HTML
from dotenv import load_dotenv
load_dotenv()
```

```python

client = OpenAI(
	api_key=os.getenv("OPENAI_API_KEY"),
)
```

```python
def load_data(file_path):
    return pd.read_csv(file_path)
```

```python
def preprocess_data(df):
    # Remove any empty rows or columns
    df = df.dropna(how='all').dropna(axis=1, how='all')
    return df
```

```python
def generate_summary(data):
    prompt = f"""
    
You will be given data tables that summarize treatment-emergent adverse events (AEs) reported in a clinical trial comparing varenicline to placebo. Your task is to analyze the cardiovascular adverse events data and relate it to the study's objectives.

Context: The study aimed to evaluate the safety and efficacy of varenicline in comparison to placebo in subjects with cardiovascular disease. Adverse events were monitored throughout the treatment period, which lasted 12 weeks, followed by a 40-week follow-up.

Summarize the total number of subjects in both treatment groups and the overall findings related to cardiovascular treatment-emergent adverse events (TEAEs).
Highlight any cardiovascular adverse events that occurred in at least 5 subjects in the varenicline group and were more frequent than in the placebo group.
Discuss the types of cardiovascular adverse events reported by system organ class, focusing on the most common occurrences such as angina pectoris and hypertension.
Provide the total number and percentage of subjects who experienced any cardiovascular adverse events in both treatment arms.
Identify any serious adverse events (SAEs) related to cardiovascular issues and discuss their relevance to treatment.
Analyze trends or patterns in the data, noting any specific cardiovascular AEs that are associated more frequently with varenicline compared to placebo.
Relate these findings back to the study's safety objectives, particularly concerning subjects with pre-existing cardiovascular conditions.
Conclude by assessing the overall cardiovascular safety profile of varenicline in comparison to placebo, discussing the implications for its use in smoking cessation therapy among patients with cardiovascular disease.
    Input tables:

    {data.to_string()}

    Please provide a summary based on the above instructions and data.
    """

    response = client.chat.completions.create(
        model="gpt-4",
        messages=[
            {"role": "system", "content": "You are a clinical trial data analyst specializing in summarizing adverse event data."},
            {"role": "user", "content": prompt}
        ],
        max_tokens=1000
    )

    return response.choices[0].message.content
```

```python
def generate_summary(data):
    prompt = f"""
    
You will be given data tables that summarize treatment-emergent adverse events (AEs) reported in a clinical trial comparing varenicline to placebo. Your task is to analyze the cardiovascular adverse events data and relate it to the study's objectives.

Context: The study aimed to evaluate the safety and efficacy of varenicline in comparison to placebo in subjects with cardiovascular disease. Adverse events were monitored throughout the treatment period, which lasted 12 weeks, followed by a 40-week follow-up.

Summarize the total number of subjects in both treatment groups and the overall findings related to cardiovascular treatment-emergent adverse events (TEAEs).
Highlight any cardiovascular adverse events that occurred in at least 5 subjects in the varenicline group and were more frequent than in the placebo group.
Discuss the types of cardiovascular adverse events reported by system organ class, focusing on the most common occurrences such as angina pectoris and hypertension.
Provide the total number and percentage of subjects who experienced any cardiovascular adverse events in both treatment arms.
Identify any serious adverse events (SAEs) related to cardiovascular issues and discuss their relevance to treatment.
Analyze trends or patterns in the data, noting any specific cardiovascular AEs that are associated more frequently with varenicline compared to placebo.
Relate these findings back to the study's safety objectives, particularly concerning subjects with pre-existing cardiovascular conditions.
Conclude by assessing the overall cardiovascular safety profile of varenicline in comparison to placebo, discussing the implications for its use in smoking cessation therapy among patients with cardiovascular disease.

    Input tables:

    {data.to_string()}

    Please provide a summary based on the above instructions and data.
    """

    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": "You are a clinical trial data analyst specializing in summarizing adverse event data."},
            {"role": "user", "content": prompt}
        ],
        max_tokens=1000
    )

  #  return response.choices[0].message.content
    summary = response.choices[0].message.content
    
    # Clean up the summary to remove hashtags and stars
    clean_summary = summary.replace('*', '').replace('#', '')
    
    return clean_summary
```

```python

file_path = 'adverse data2.csv'
df = load_data(file_path)
processed_df = preprocess_data(df)
```
